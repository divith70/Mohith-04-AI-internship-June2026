```python
#arithmetic operator
num1=int(input("enter the first number"))
num2=int(input("enter the second number"))
print("addition",num1+num2)
print("subtraction",num1-num2)
print("multiplication",num1*num2)
print("division",num1/num2)
print("modulus",num1%num2)
print("exponent",num1**num2)
print("floor division",num1//num2)1
```

```python
#relational operator
num1=int(input("enter the first number"))
num2=int(input("enter the second number"))
print(num1==num2)
print("num1 > num2 is", num1 > num2)
print("num1 < num2 is", num1 < num2)
print("num1 == num2 is", num1 == num2)
print("num1 != num2 is", num1 != num2)
print("num1 >= num2 is", num1 >= num2)
print("num1 <= num2 is", num1 <= num2)
```

```python
#logical operator
num1=20
num2=30
print(num1>10 and num2<30)
print(num1>10 or num2<30)
print(not( num1>10 and num2<30))
```

```python
#special operator-identity
cls1="gowtham","meera","mohith","nivas"
cls2="gowtham","meera","dinesh","nivas"
print(cls1 is cls2)
print(cls1 is not cls2)
```

```python
#special operators-membership
cls="gowtham","mohith","sankar"
print("mohith" in cls)
print("dinesh" not in cls)
print("sankar" not in cls)
```

```python
#control flow-elif
marks=int(input("enter the marks :"))
if marks>=90:
  print("grade A")
elif marks>=75:
  print("grade B")
elif marks>=50:
  print("grade C")
else:
  print("fail")
```

```python
#control flow-nested if
num=20
if num<30:
  print("num is less than 30")
  if num % 2==0:
    print("num is even")
  else:
    print("num is odd")
else:
  print("num is greater than 30")
```
