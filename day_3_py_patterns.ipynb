```python
# 1
n = 5
for i in range(1, n + 1):
    for j in range(1,n+1):
        print(i, end=" ")
    print()
```

```python
# 2
n=5
for i in range(n,0,-1):
  for j in range(i):
    print("*",end=" ")
  print()
```

```python
# 3
n=5
for i in range(n):
  for j in range(n):
    if i==0:
      print("*",end=" ")
    elif i==n-1:
      print("*",end=" ")
    elif j==0:
      print( "*",end=" ")
    elif j==n-1:
      print("*",end=" ")

    else :
      print(" ",end=" ")
  print()
```
