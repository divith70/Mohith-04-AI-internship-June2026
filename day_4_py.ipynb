```python
# string methods
n="divi"
print(n.find('i'))
print(n.rfind("i"))
print(n.index("v"))
print(n.count("v"))
print(n.replace("i","$"))
print(n.startswith("d"))
print(n.endswith("h"))
```

```python
# testing string contents


# Example 1:
s1 = "HelloPython123"
print(f"'{s1}' is alphanumeric: {s1.isalnum()}")
print(f"'{s1}' is alphabetic: {s1.isalpha()}")
print(f"'{s1}' is digit: {s1.isdigit()}")
print(f"'{s1}' is lowercase: {s1.islower()}")
print(f"'{s1}' is uppercase: {s1.isupper()}")
print(f"'{s1}' is title case: {s1.istitle()}")
print(f"'{s1}' is space: {s1.isspace()}")



```

```python
# string formating
name="mohith"
age=18
print(f"my name is {name} and my age is {age}")
print("my name is {} and my age is {}".format(name,age))
print("my name is {n} and my age is {a}".format(n=name,a=age))
```

```python
# list
l=[1,2,3,4,5,6]
print(type(l))
for i in l:
  print(i)
print(len(l))
print(l[3])
l.remove(4)
print(l)
l.reverse()
print(l)
l.sort()
print(l)
print(l.pop())
```

```python
#set
s={1,2,3,4,5,6}
print(type(s))
s.clear()
print(s )
p={1,2,3,4,5}
b={5,6,7,8,9}
p.discard(5)
print((p))
print(p.difference(b))
p.difference_update(b)
print(p)
a={1,2,3,4,5}
c={5,6,7,8,9}
print(a.intersection(c))
a.intersection_update(c)
print(a)
print(c.isdisjoint(a))
print(c.issubset(a))
print(a.issuperset(c))
print(a.symmetric_difference(c))
a.symmetric_difference_update(c)
print(a)
```
